## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression linéaire (*Linear Regression*, moindres carrés ordinaires). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique à partir de plusieurs variables explicatives.

**Fonctionnement :** en entrée, on a une matrice $X$ de $n$ lignes et $p$ colonnes et un vecteur cible $y$. L'apprentissage calcule les coefficients qui placent l'hyperplan au plus près de tous les points. En sortie, le modèle donne une prédiction pour toute nouvelle observation.

**Formule :** $\hat{y}_i = \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij}$, avec le critère $J(\beta) = \frac{1}{2n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$. Ici $n$ est le nombre d'observations, $p$ le nombre de variables, $x_{ij}$ la valeur de la variable $j$ pour l'observation $i$, $y_i$ la vraie valeur, $\hat{y}_i$ la prédiction, $\beta_0$ la constante et $\beta_j$ le coefficient de la variable $j$.

**Paramètres / hyperparamètres :** les paramètres appris sont $\beta_0,\dots,\beta_p$. Il n'y a pratiquement pas d'hyperparamètre, seulement `fit_intercept` (garder ou non la constante).

**Avantages :** rapide à calculer, et chaque coefficient s'interprète (variation de $y$ quand $x_j$ augmente d'une unité, les autres variables restant fixes). **Limites :** sensible aux valeurs extrêmes, et incapable de représenter une relation courbe ; les coefficients deviennent instables si les variables sont très corrélées.

### Schéma explicatif

```python
# Schéma 1 : droite des moindres carrés et résidus (une seule variable, données jouet)
from sklearn.linear_model import LinearRegression
rng1 = np.random.default_rng(GRAINE)
x1 = rng1.uniform(0, 10, 25)
y1 = 2 + 1.5 * x1 + rng1.normal(0, 2, 25)
m1 = LinearRegression().fit(x1.reshape(-1, 1), y1)
pred1 = m1.predict(x1.reshape(-1, 1))
xs = np.linspace(0, 10, 100).reshape(-1, 1)
plt.figure(figsize=(6, 4))
plt.scatter(x1, y1, color='tab:blue', label='observations')
plt.plot(xs, m1.predict(xs), color='tab:red', label='droite ajustée')
plt.vlines(x1, y1, pred1, color='gray', alpha=.6, label='résidus (erreurs)')
plt.xlabel('x (variable explicative)'); plt.ylabel('y (cible)')
plt.title('Régression linéaire : minimiser les carrés des résidus')
plt.legend(); plt.show()
```

**Comment lire le schéma :** les points bleus sont les observations et la droite rouge est le modèle appris. Chaque trait gris vertical est un résidu, c'est-à-dire l'écart entre le point réel et la prédiction. L'algorithme choisit la droite pour que la **somme des carrés** de ces traits soit la plus petite possible. Les données sont synthétiques et servent uniquement à illustrer.

### Contextes d'utilisation

- **Cas 1 (gestion / marketing) :** prévoir le chiffre d'affaires mensuel d'un magasin à partir du budget publicitaire, du nombre de visiteurs et du prix moyen. **Cas 2 :** estimer le prix d'un logement à partir de sa surface et de son nombre de pièces.
- **Données et prérequis :** variables numériques (ou catégorielles transformées en indicatrices), cible continue, plus d'observations que de variables, et relation à peu près linéaire entre les variables et la cible.
- **Cas peu approprié :** prédire si un client va faire défaut (cible oui/non), car la droite peut sortir de l'intervalle [0, 1] ; on utilise alors la régression logistique. Elle convient mal aussi à une relation nettement courbe.
- **Métriques :** MAE, RMSE et R² sur un jeu de test, complétés par l'étude des résidus.

### Code fourni (conservé) et sortie

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xr_ent` (210 lignes × 5 variables) et `yr_ent` pour l'apprentissage, puis `Xr_test` (70 lignes) pour mesurer l'erreur. Les données viennent de `make_regression`, donc elles sont **synthétiques**. **Hyperparamètres :** aucun réglé (`fit_intercept=True` par défaut). **Sortie :** la MAE sur le test, soit 9,354 avec scikit-learn 1.9.1 (à relire dans votre exécution).

**Ce que la sortie permet de dire :** en moyenne, la prédiction s'écarte d'environ 9,4 unités de la vraie valeur, alors que l'écart-type de `y` sur le test est d'environ 97,6. L'erreur est donc petite devant l'étendue de la cible.

**Ce qu'elle ne permet pas de dire :** une MAE seule ne suffit pas pour juger un modèle, d'où le tableau avec RMSE et R² plus bas. De plus, `make_regression` fabrique `y` comme une combinaison linéaire des variables plus du bruit ; le modèle linéaire a donc un avantage structurel qui n'existerait pas forcément sur des données réelles. Enfin, il n'y a qu'un seul découpage train/test.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression Ridge (*Ridge regression*, régularisation L2). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique en gardant des coefficients de taille raisonnable, pour limiter le surapprentissage.

**Fonctionnement :** on part du modèle linéaire, mais on ajoute au critère une pénalité sur la taille des coefficients. En entrée : $X$ standardisée et $y$. L'apprentissage cherche le meilleur compromis entre l'erreur et la taille des coefficients. En sortie, on obtient une prédiction comme pour la régression linéaire.

**Formule (version de scikit-learn) :** $J(\beta) = \lVert y - X\beta \rVert_2^2 + \alpha \sum_{j=1}^{p}\beta_j^2$. Ici $X$ est la matrice $n \times p$, $y$ le vecteur des cibles, $\beta$ le vecteur des coefficients (la constante n'est pas pénalisée), $\lVert\cdot\rVert_2$ la norme euclidienne et $\alpha \ge 0$ la force de la pénalité. Le cours écrit la même idée avec un facteur $1/n$ et la lettre $\lambda$ ; les deux sont reliés par $\lambda = \alpha / n$.

**Paramètres / hyperparamètres :** les paramètres appris sont les $\beta_j$ ; l'hyperparamètre est $\alpha$ (ici 1,0), à choisir par validation. On standardise d'abord les variables, sinon la pénalité punit davantage les variables exprimées en petites unités.

**Avantages :** stable quand les variables sont corrélées, et réduit la variance du modèle. **Limites :** ne met pas les coefficients exactement à zéro (pas de sélection de variables), et les coefficients sont volontairement tirés vers 0, ce qui les rend un peu biaisés.

### Schéma explicatif

```python
# Schéma 2 : chemin des coefficients de Ridge quand alpha augmente
from sklearn.linear_model import Ridge
Xs = StandardScaler().fit_transform(Xr_ent)        # standardisation sur l'entraînement seulement
alphas = np.logspace(-2, 4, 40)
chemins = np.array([Ridge(alpha=a).fit(Xs, yr_ent).coef_ for a in alphas])
plt.figure(figsize=(6, 4))
for j in range(chemins.shape[1]):
    plt.plot(alphas, chemins[:, j], label='variable %d' % (j + 1))
plt.xscale('log'); plt.axhline(0, color='black', lw=.8)
plt.xlabel('alpha (échelle logarithmique)'); plt.ylabel('coefficient appris')
plt.title('Ridge : les coefficients se contractent quand alpha augmente')
plt.legend(); plt.show()
```

**Comment lire le schéma :** chaque courbe est un coefficient. À gauche (alpha très petit) on retrouve les coefficients de la régression linéaire. En allant vers la droite, la pénalité augmente et **tous les coefficients se rapprochent de zéro** en même temps, sans jamais être annulés brutalement. Le bon alpha se situe quelque part au milieu et se choisit par validation.

### Contextes d'utilisation

- **Cas 1 (finance / gestion) :** prévoir les ventes ou le rendement d'un actif à partir d'indicateurs économiques très corrélés (inflation, taux, PIB). **Cas 2 :** estimer une consommation d'énergie à partir de plusieurs capteurs qui mesurent presque la même chose.
- **Données et prérequis :** variables numériques, souvent nombreuses et corrélées, standardisées avant l'ajustement ; un $\alpha$ choisi par validation croisée.
- **Cas peu approprié :** quand on veut un modèle avec **peu de variables retenues** (le Lasso est plus adapté), ou quand la relation est fortement non linéaire.
- **Métriques :** MAE, RMSE et R², calculés en validation croisée pour choisir $\alpha$, puis sur le test final.

### Code fourni (conservé) et sortie

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

**Entrées :** les mêmes `Xr_ent` / `yr_ent` et `Xr_test` que pour l'algorithme 1, passées dans un `Pipeline` qui standardise puis applique Ridge. **Hyperparamètre :** `alpha=1.0`, qui n'a pas été réglé. **Sortie :** la MAE sur le test, soit 9,368 avec scikit-learn 1.9.1.

**Ce que la sortie permet de dire :** le résultat est presque identique à celui de la régression linéaire (9,354). Cela s'explique par le contexte : 210 observations pour seulement 5 variables, peu de corrélation entre elles, et un alpha faible. Il y a peu de variance à corriger, donc Ridge ne change presque rien.

**Ce qu'elle ne permet pas de dire :** on ne peut pas conclure que Ridge est « moins bon » : une différence de l'ordre de 0,01 est négligeable et vient d'un seul découpage. On ne peut pas non plus juger l'intérêt de Ridge, car il se voit surtout avec des variables très corrélées ou peu d'observations, ce qui n'est pas le cas ici. Il faudrait comparer plusieurs valeurs de alpha en validation croisée.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression Lasso (*Lasso regression*, régularisation L1). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique tout en sélectionnant automatiquement les variables utiles.

**Fonctionnement :** comme Ridge, on ajoute une pénalité au modèle linéaire, mais ici elle porte sur la **valeur absolue** des coefficients. Cette forme de pénalité peut ramener certains coefficients exactement à zéro. En entrée : $X$ standardisée et $y$ ; en sortie : un modèle linéaire où seules quelques variables ont un coefficient non nul.

**Formule (version de scikit-learn) :** $J(\beta) = \frac{1}{2n}\lVert y - X\beta\rVert_2^2 + \alpha\sum_{j=1}^{p}\lvert\beta_j\rvert$. Ici $n$ est le nombre d'observations, $X$ la matrice des variables, $y$ la cible, $\beta_j$ les coefficients, $\alpha \ge 0$ la force de la pénalité et $\lvert\cdot\rvert$ la valeur absolue.

**Paramètres / hyperparamètres :** paramètres appris : les $\beta_j$ ; hyperparamètres : $\alpha$ (ici 0,1) et `max_iter` (10 000), qui limite le nombre d'itérations de l'algorithme de calcul.

**Avantages :** sélection de variables intégrée et modèle plus simple à expliquer. **Limites :** instable quand plusieurs variables sont fortement corrélées (il en choisit une un peu au hasard), et les coefficients conservés sont biaisés vers 0.

### Schéma explicatif

```python
# Schéma 3 : chemin des coefficients du Lasso quand alpha augmente
from sklearn.linear_model import Lasso
Xs = StandardScaler().fit_transform(Xr_ent)
alphas = np.logspace(-2, 2.4, 40)
chemins = np.array([Lasso(alpha=a, max_iter=10000).fit(Xs, yr_ent).coef_ for a in alphas])
plt.figure(figsize=(6, 4))
for j in range(chemins.shape[1]):
    plt.plot(alphas, chemins[:, j], label='variable %d' % (j + 1))
plt.xscale('log'); plt.axhline(0, color='black', lw=.8)
plt.xlabel('alpha (échelle logarithmique)'); plt.ylabel('coefficient appris')
plt.title('Lasso : les coefficients tombent à zéro les uns après les autres')
plt.legend(); plt.show()
```

**Comment lire le schéma :** chaque courbe est un coefficient. Quand alpha augmente, les coefficients diminuent, mais contrairement à Ridge ils **touchent zéro** : la variable correspondante sort alors du modèle. Les variables qui s'annulent en dernier sont celles que le Lasso juge les plus utiles. Dans mon exemple, les 5 variables servent toutes à fabriquer `y`, donc les coefficients ne s'annulent que pour des alphas assez grands.

### Contextes d'utilisation

- **Cas 1 (finance) :** choisir quelques ratios financiers vraiment utiles parmi des dizaines pour prévoir la rentabilité d'une entreprise. **Cas 2 :** repérer, parmi de nombreuses caractéristiques d'un produit, celles qui expliquent le mieux le prix.
- **Données et prérequis :** nombreuses variables numériques standardisées, avec l'idée qu'une partie seulement est utile ; un $\alpha$ choisi par validation croisée.
- **Cas peu approprié :** quand des groupes de variables très corrélées sont **toutes** importantes, car le Lasso en garde une seule de façon instable. Ridge est alors plus raisonnable.
- **Métriques :** MAE, RMSE, R², et nombre de coefficients non nuls pour juger la parcimonie.

### Code fourni (conservé) et sortie

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xr_ent` / `yr_ent` pour l'ajustement et `Xr_test` pour l'évaluation, avec standardisation dans le `Pipeline`. **Hyperparamètres :** `alpha=0.1` et `max_iter=10000`, non réglés. **Sortie :** le R² sur le test, qui vaut 0,986 avec scikit-learn 1.9.1.

**Ce que la sortie permet de dire :** le modèle explique environ 98,6 % de la variance de la cible sur le test. Les coefficients appris sont tous non nuls : le Lasso n'a éliminé aucune variable, ce qui est cohérent puisque `make_regression` utilise ici les 5 variables pour construire `y`.

**Ce qu'elle ne permet pas de dire :** le R² seul ne dit rien de l'erreur en unités de `y` (d'où la MAE et la RMSE du tableau plus bas). Sur ces données, on ne peut pas voir la vraie force du Lasso, qui est de supprimer des variables inutiles, puisqu'il n'y en a aucune. Une comparaison sérieuse nécessiterait des données réelles avec beaucoup de variables.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression polynomiale (*Polynomial regression*). **Famille :** régression supervisée. **Objectif :** modéliser une relation courbe entre les variables et la cible.

**Fonctionnement :** on commence par fabriquer de nouvelles variables (puissances et produits des variables d'origine), puis on applique un modèle linéaire sur ces variables enrichies. Avec 5 variables et `degree=2`, on passe à 20 variables : les 5 d'origine, leurs 5 carrés et les 10 produits deux à deux. En sortie, on obtient une courbe (ou une surface) au lieu d'une droite.

**Formule (une variable) :** $\hat{y} = \beta_0 + \beta_1 x + \beta_2 x^2 + \dots + \beta_d x^d$. Ici $x$ est la variable d'entrée, $d$ le degré maximal, $\beta_j$ les coefficients appris et $\hat{y}$ la valeur estimée. Le modèle reste **linéaire en $\beta$** même s'il est courbe en $x$.

**Paramètres / hyperparamètres :** paramètres appris : les $\beta_j$ ; hyperparamètres : le degré `degree` (ici 2) et, dans ce code, `alpha=10` de la pénalité Ridge qui freine les coefficients.

**Avantages :** capture des courbures et des interactions tout en gardant un modèle simple à ajuster. **Limites :** le nombre de variables explose avec le degré, et un degré trop élevé fait surapprendre et extrapoler de façon absurde.

### Schéma explicatif

```python
# Schéma 4 : même petit nuage de points, trois degrés de polynôme (données jouet)
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
x_app = x_jouet[::4].reshape(-1, 1) / 3          # 15 points, ramenés dans [-1, 1]
y_app = y_jouet[::4]
grille = x_grille / 3
plt.figure(figsize=(6.5, 4))
plt.scatter(x_app * 3, y_app, color='black', zorder=3, label='15 points d\'apprentissage')
for deg, couleur in [(1, 'tab:blue'), (3, 'tab:green'), (12, 'tab:red')]:
    mod = make_pipeline(PolynomialFeatures(degree=deg), LinearRegression()).fit(x_app, y_app)
    plt.plot(x_grille, mod.predict(grille), color=couleur, label='degré %d' % deg)
plt.ylim(-3, 3); plt.xlabel('x'); plt.ylabel('y')
plt.title('Polynomiale : sous-apprentissage, bon degré, surapprentissage')
plt.legend(); plt.show()
```

**Comment lire le schéma :** les points noirs sont les données d'apprentissage. Le degré 1 (bleu) est une droite : trop simple pour suivre la courbe, c'est le **sous-apprentissage**. Le degré 3 (vert) suit mieux la forme générale (une descente puis une remontée), même s'il s'écarte des points à droite. Le degré 12 (rouge) passe très près des points d'apprentissage mais oscille fortement entre eux et part à l'extrême sur les bords (la courbe sort du cadre) : c'est le **surapprentissage**. Le degré est donc un hyperparamètre à régler sur la validation.

### Contextes d'utilisation

- **Cas 1 (marketing / gestion) :** modéliser l'effet du prix sur la demande, quand la demande baisse de plus en plus vite avec le prix. **Cas 2 :** décrire l'évolution non linéaire du rendement d'une culture selon la quantité d'engrais.
- **Données et prérequis :** peu de variables (le nombre de termes grossit vite), variables standardisées, idéalement une pénalisation (Ridge) pour garder le modèle stable.
- **Cas peu approprié :** prévoir au-delà de la plage des données, car un polynôme extrapole mal ; il convient aussi mal à des données avec beaucoup de variables ou de fortes ruptures.
- **Métriques :** MAE, RMSE, R² sur des données de validation, avec comparaison à un degré 1 de référence.

### Code fourni (conservé) et sortie

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xr_ent` / `yr_ent` et `Xr_test`. Le `Pipeline` standardise, ajoute les termes de degré 2 (sans constante, `include_bias=False`), puis applique un Ridge. **Hyperparamètres :** `degree=2` et `alpha=10`, non réglés. **Sortie :** la MAE sur le test, soit 10,897 avec scikit-learn 1.9.1.

**Ce que la sortie permet de dire :** l'erreur est plus élevée que celle de la régression linéaire (9,354). Ici, les 15 termes ajoutés (carrés et produits) n'apportent aucune information utile puisque la vraie relation est linéaire, et ils ajoutent surtout du bruit à ajuster. La pénalité Ridge limite les dégâts, ce qui explique que l'écart reste modéré.

**Ce qu'elle ne permet pas de dire :** on ne peut pas en déduire que la régression polynomiale est moins bonne en général : sur des données réellement courbes, elle serait avantagée. Ce résultat vient du fait que les données synthétiques sont linéaires par construction, et d'un seul découpage.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Arbre de régression (*Regression tree*). **Famille :** régression supervisée. **Objectif :** prédire une valeur numérique en découpant l'espace des variables en zones simples.

**Fonctionnement :** en entrée, $X$ et $y$. L'apprentissage teste des coupures du type « variable $j$ inférieure à un seuil » et retient celle qui rend les valeurs de $y$ les plus homogènes dans chaque morceau. On répète cela dans chaque morceau, jusqu'à la profondeur maximale. En sortie, une nouvelle observation tombe dans une feuille et reçoit la moyenne des $y$ de cette feuille.

**Formule :** $\hat{y}(x) = \frac{1}{\lvert L(x)\rvert}\sum_{i \in L(x)} y_i$. Ici $L(x)$ est l'ensemble des observations d'entraînement qui se trouvent dans la même feuille que $x$, $\lvert L(x)\rvert$ leur nombre et $y_i$ leur valeur. Le critère de coupure de scikit-learn est la réduction de l'erreur quadratique à l'intérieur des feuilles.

**Paramètres / hyperparamètres :** paramètres appris : les variables et seuils de coupure, et la valeur de chaque feuille ; hyperparamètres : `max_depth` (ici 5), `min_samples_leaf`, etc.

**Avantages :** gère les relations non linéaires et se lit comme une suite de règles, sans besoin de standardiser. **Limites :** instable (un petit changement dans les données peut changer l'arbre) et prédictions constantes par morceaux, donc mauvaise extrapolation.

### Schéma explicatif

```python
# Schéma 5 : un arbre de régression découpe l'axe en paliers (données jouet)
from sklearn.tree import DecisionTreeRegressor
plt.figure(figsize=(6.5, 4))
plt.scatter(x_jouet, y_jouet, color='black', s=15, label='observations')
for prof, couleur in [(2, 'tab:blue'), (6, 'tab:red')]:
    arbre = DecisionTreeRegressor(max_depth=prof, random_state=GRAINE).fit(x_jouet.reshape(-1, 1), y_jouet)
    plt.plot(x_grille, arbre.predict(x_grille), color=couleur, label='max_depth = %d' % prof)
plt.xlabel('x'); plt.ylabel('y')
plt.title('Arbre de régression : une prédiction constante par feuille')
plt.legend(); plt.show()
```

**Comment lire le schéma :** chaque palier horizontal correspond à une feuille de l'arbre, avec comme valeur la moyenne des points qu'elle contient. Avec `max_depth=2` (bleu) il y a au plus 4 paliers : la courbe est très grossière. Avec `max_depth=6` (rouge), il y a beaucoup plus de paliers et la courbe suit mieux les points, mais elle commence à copier aussi le bruit. La profondeur règle donc la complexité du modèle.

### Contextes d'utilisation

- **Cas 1 (logistique) :** estimer le coût de livraison selon la distance, le poids et la zone, avec des seuils nets (par exemple un tarif qui change au-delà de 20 kg). **Cas 2 :** prévoir le prix d'un bien selon des règles comme « surface > 100 m² et quartier X ».
- **Données et prérequis :** variables numériques ou catégorielles encodées, pas de standardisation nécessaire ; la profondeur et la taille minimale des feuilles doivent être contrôlées.
- **Cas peu approprié :** une relation lisse et régulière, comme ici une relation linéaire : l'arbre doit l'approcher avec des marches d'escalier. Il convient mal aussi pour prévoir hors de la plage des données.
- **Métriques :** MAE, RMSE et R² sur la validation et le test ; la lisibilité des règles est un critère qualitatif.

### Code fourni (conservé) et sortie

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xr_ent` / `yr_ent` pour l'apprentissage et `Xr_test` pour l'évaluation. **Hyperparamètres :** `max_depth=5` et `random_state=GRAINE` (pour reproduire le résultat) ; ils n'ont pas été réglés. **Sortie :** la MAE sur le test, soit 37,717 avec scikit-learn 1.9.1.

**Ce que la sortie permet de dire :** l'erreur est environ quatre fois plus grande que celle de la régression linéaire (9,354). C'est logique : la vraie relation est linéaire, et un arbre de profondeur 5 (au maximum 32 feuilles) doit la reconstruire avec des paliers, en s'appuyant sur seulement 210 observations, donc peu de points par feuille.

**Ce qu'elle ne permet pas de dire :** cela ne prouve pas que les arbres sont de mauvais modèles. Sur des données avec des seuils, des interactions ou des relations non linéaires, ils peuvent faire mieux que la régression linéaire. Ce résultat dépend de la nature synthétique et linéaire des données, d'un seul découpage et d'une profondeur non réglée.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Forêt aléatoire pour la régression (*Random Forest Regressor*). **Famille :** régression supervisée (méthode d'ensemble). **Objectif :** prédire une valeur numérique en moyennant de nombreux arbres, pour réduire l'instabilité d'un arbre seul.

**Fonctionnement :** en entrée, $X$ et $y$. L'apprentissage fabrique plusieurs échantillons tirés **avec remise** (bootstrap) et entraîne un arbre sur chacun. En sortie, la prédiction est la moyenne des prédictions de tous les arbres.

**Formule :** $\hat{y}(x) = \frac{1}{B}\sum_{b=1}^{B} T_b(x)$. Ici $B$ est le nombre d'arbres, $T_b(x)$ la prédiction de l'arbre $b$ pour l'entrée $x$ et $\hat{y}(x)$ la moyenne.

**Paramètres / hyperparamètres :** paramètres appris : les coupures et les valeurs des feuilles de chaque arbre ; hyperparamètres : `n_estimators` (ici 80), `max_depth`, `max_features`. Attention : dans scikit-learn, `RandomForestRegressor` utilise par défaut **toutes** les variables à chaque coupure (`max_features=1.0`), donc ici la diversité entre arbres vient surtout du bootstrap, alors que le cours parle aussi d'un tirage de sous-ensembles de variables.

**Avantages :** réduit fortement la variance d'un arbre seul et demande peu de réglages. **Limites :** modèle plus volumineux et moins interprétable, et il ne peut pas prédire des valeurs en dehors de l'intervalle des $y$ observés.

### Schéma explicatif

```python
# Schéma 6 : 20 arbres sur des échantillons bootstrap, puis leur moyenne (données jouet)
from sklearn.tree import DecisionTreeRegressor
rng6 = np.random.default_rng(GRAINE)
predictions = []
plt.figure(figsize=(6.5, 4))
for b in range(20):
    tirage = rng6.integers(0, len(x_jouet), len(x_jouet))      # bootstrap : tirage avec remise
    arbre = DecisionTreeRegressor(max_depth=6, random_state=GRAINE + b)
    arbre.fit(x_jouet[tirage].reshape(-1, 1), y_jouet[tirage])
    predictions.append(arbre.predict(x_grille))
    plt.plot(x_grille, predictions[-1], color='gray', alpha=.35, lw=1,
             label='un arbre (20 au total)' if b == 0 else None)
plt.plot(x_grille, np.mean(predictions, axis=0), color='tab:red', lw=2.5, label='moyenne = forêt')
plt.scatter(x_jouet, y_jouet, color='black', s=12, label='observations')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Forêt aléatoire : la moyenne lisse les arbres individuels')
plt.legend(); plt.show()
```

**Comment lire le schéma :** chaque courbe grise est un arbre entraîné sur un échantillon bootstrap différent ; ils se ressemblent mais diffèrent dans les détails, car chacun a vu des données un peu différentes. La courbe rouge est leur **moyenne**, c'est la prédiction de la forêt : elle est plus régulière que chaque arbre pris séparément. C'est ce lissage qui réduit la variance.

### Contextes d'utilisation

- **Cas 1 (commerce international) :** prévoir la demande export d'un produit à partir du prix, de la saison, de la taille du marché et du taux de change. **Cas 2 :** estimer le temps de livraison selon la distance, le transporteur et la météo.
- **Données et prérequis :** données tabulaires numériques ou encodées, valeurs manquantes à traiter, pas de standardisation nécessaire ; un nombre d'arbres suffisant.
- **Cas peu approprié :** une tendance qui continue au-delà des données connues (par exemple des ventes qui croissent chaque année), car la forêt ne dépasse jamais les valeurs vues. Peu adaptée aussi quand l'interprétabilité de chaque prédiction est exigée.
- **Métriques :** MAE, RMSE et R² sur le test, et erreur « out-of-bag » pour estimer la généralisation.

### Code fourni (conservé) et sortie

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xr_ent` / `yr_ent` pour l'apprentissage et `Xr_test` pour l'évaluation. **Hyperparamètres :** `n_estimators=80` et `random_state=GRAINE` ; la profondeur est laissée libre et rien n'est réglé. **Sortie :** la MAE sur le test, soit 24,374 avec scikit-learn 1.9.1.

**Ce que la sortie permet de dire :** par rapport à l'arbre unique de l'exemple 5 (37,717), l'erreur baisse nettement. Moyenner de nombreux arbres réduit donc bien la variance. La forêt reste pourtant plus loin du résultat de la régression linéaire (9,354), car les arbres approchent toujours une relation linéaire avec des paliers.

**Ce qu'elle ne permet pas de dire :** on ne sait pas, avec cette seule sortie, si 80 arbres suffisent ni si d'autres réglages feraient mieux. La comparaison avec le modèle linéaire dépend de la nature linéaire des données synthétiques et ne vaut pas pour des données réelles. Comme il y a un seul découpage, l'écart entre modèles n'a pas de mesure d'incertitude.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Gradient boosting pour la régression (*Gradient Boosting Regressor*) ; l'exemple utilise `HistGradientBoostingRegressor`. **Famille :** régression supervisée (méthode d'ensemble séquentielle). **Objectif :** prédire une valeur numérique en ajoutant de petits arbres qui corrigent les erreurs des précédents.

**Fonctionnement :** en entrée, $X$ et $y$. On part d'une prédiction simple (souvent la moyenne de $y$), on calcule l'erreur restante de chaque observation, on entraîne un petit arbre pour la prédire, puis on ajoute ce arbre, réduit par un coefficient, au modèle. On répète ces étapes. En sortie, la prédiction est la somme de tous les arbres ajoutés.

**Formule :** $F_m(x) = F_{m-1}(x) + \eta\, h_m(x)$. Ici $F_m$ est le modèle après $m$ étapes, $h_m$ le nouvel arbre qui apprend les erreurs, $\eta$ le taux d'apprentissage (la taille du pas) et $x$ les données d'entrée.

**Paramètres / hyperparamètres :** paramètres appris : la structure et les valeurs des arbres ; hyperparamètres : `max_iter` (nombre d'arbres, ici 80), `learning_rate` ($\eta$, 0,1 par défaut) et la profondeur des arbres. La version « Hist » regroupe les valeurs des variables en classes (histogrammes) pour aller plus vite ; ce n'est pas exactement le `GradientBoostingRegressor` classique.

**Avantages :** souvent très performant sur des tableaux de données et rapide avec la version Hist. **Limites :** plus sensible aux réglages que la forêt, et risque de surapprentissage si on ajoute trop d'arbres.

### Schéma explicatif

```python
# Schéma 7 : le modèle s'affine quand on ajoute des arbres (données jouet)
from sklearn.ensemble import HistGradientBoostingRegressor
fig, axes = plt.subplots(1, 3, figsize=(12, 3.6), sharey=True)
for ax, nb_arbres in zip(axes, [1, 5, 50]):
    mod = HistGradientBoostingRegressor(max_iter=nb_arbres, learning_rate=0.3,
                                        min_samples_leaf=5, random_state=GRAINE)
    mod.fit(x_jouet.reshape(-1, 1), y_jouet)
    ax.scatter(x_jouet, y_jouet, color='black', s=12)
    ax.plot(x_grille, mod.predict(x_grille), color='tab:red', lw=2)
    ax.set_title('%d arbre(s) ajouté(s)' % nb_arbres); ax.set_xlabel('x')
axes[0].set_ylabel('y')
fig.suptitle('Gradient boosting : chaque arbre corrige les erreurs restantes', y=1.06); plt.show()
```

**Comment lire le schéma :** de gauche à droite, le même modèle avec de plus en plus d'arbres (1, 5, puis 50). Les points noirs sont les observations et la courbe rouge est la prédiction du modèle. Avec peu d'arbres, la courbe rouge est encore éloignée des points ; à chaque arbre ajouté, elle se rapproche de la forme de la courbe. Après beaucoup d'étapes, elle colle de près aux points, avec un risque de copier aussi le bruit. Le nombre d'arbres est donc un hyperparamètre à régler.

### Contextes d'utilisation

- **Cas 1 (marketing / commerce) :** prévoir les ventes d'un produit à partir de prix, promotions, saisonnalité et historique. **Cas 2 :** prévoir le coût d'une réparation ou d'un sinistre à partir de caractéristiques variées.
- **Données et prérequis :** données tabulaires de taille moyenne à grande, variables numériques ou encodées, validation soignée pour régler le nombre d'arbres et le taux d'apprentissage.
- **Cas peu approprié :** un très petit jeu de données, où le risque de surapprentissage est fort, ou un contexte où chaque prédiction doit être expliquée simplement à un client.
- **Métriques :** MAE, RMSE et R² sur la validation et sur le test ; courbe d'erreur selon le nombre d'itérations.

### Code fourni (conservé) et sortie

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xr_ent` / `yr_ent` pour l'apprentissage et `Xr_test` pour l'évaluation. **Hyperparamètres :** `max_iter=80` et `random_state=GRAINE` ; les autres valeurs (dont le taux d'apprentissage) sont celles par défaut, et rien n'est réglé. **Sortie :** la MAE sur le test, soit 23,266 avec scikit-learn 1.9.1.

**Ce que la sortie permet de dire :** l'erreur est un peu plus faible que celle de la forêt aléatoire (24,374) et bien plus faible que celle de l'arbre seul (37,717). Cela illustre l'intérêt de corriger les erreurs étape par étape. Elle reste cependant plus élevée que celle de la régression linéaire (9,354).

**Ce qu'elle ne permet pas de dire :** l'écart avec la forêt est modeste et vient d'un seul découpage : on ne peut pas affirmer que le boosting est meilleur. Comme les données synthétiques sont linéaires, le résultat favorise les modèles linéaires et ne dit pas ce qui se passerait sur des données réelles. Le nombre d'itérations (80) n'est pas choisi sur une validation.

## Comparaison des algorithmes de régression 1 à 7 (données synthétiques)

Le TAF demande MAE, RMSE et R² pour la régression. Je recalcule donc les trois métriques pour les 7 modèles avec **le même découpage train/test**, et je mesure aussi le temps d'ajustement pour discuter le coût de calcul. Rappel : les données viennent de `make_regression` ; ce tableau **n'est pas** un résultat sur des données réelles.

```python
import time
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor

modeles_reg = {
    '1. Régression linéaire': LinearRegression(),
    '2. Ridge': make_pipeline(StandardScaler(), Ridge(alpha=1.0)),
    '3. Lasso': make_pipeline(StandardScaler(), Lasso(alpha=.1, max_iter=10000)),
    '4. Polynomiale (deg 2)': make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=10)),
    '5. Arbre de régression': DecisionTreeRegressor(max_depth=5, random_state=GRAINE),
    '6. Forêt aléatoire': RandomForestRegressor(n_estimators=80, random_state=GRAINE),
    '7. Gradient boosting (Hist)': HistGradientBoostingRegressor(max_iter=80, random_state=GRAINE),
}
lignes = []
for nom, mod in modeles_reg.items():
    t0 = time.perf_counter()
    mod.fit(Xr_ent, yr_ent)
    duree = time.perf_counter() - t0
    pred = mod.predict(Xr_test)
    lignes.append({'Algorithme': nom,
                   'MAE': mean_absolute_error(yr_test, pred),
                   'RMSE': np.sqrt(mean_squared_error(yr_test, pred)),
                   'R²': r2_score(yr_test, pred),
                   'Temps ajustement (s)': duree})
tableau_reg = pd.DataFrame(lignes).round(3)
print(tableau_reg.to_string(index=False))
print('Écart-type de y sur le test :', round(float(yr_test.std()), 2))
```

Avec scikit-learn 1.9.1, j'obtiens les résultats suivants sur le test (les temps varient d'une machine à l'autre, donc je ne les commente pas par un chiffre) :

- Les trois modèles linéaires (1, 2, 3) sont quasiment à égalité : MAE d'environ 9,35 à 9,37, RMSE d'environ 11,5 et R² de 0,986. Leur RMSE est proche du bruit injecté dans `make_regression` (écart-type 13), ce qui montre qu'ils atteignent presque la limite possible sur ces données.
- La polynomiale de degré 2 (4) est un peu moins bonne (MAE 10,897) : elle ajoute des termes inutiles.
- L'arbre (5) est le moins bon (MAE 37,717, R² 0,742). La forêt (6) et le boosting (7) font nettement mieux que l'arbre (MAE 24,374 et 23,266, R² 0,871 et 0,885) mais restent loin des modèles linéaires.
- **Conclusion prudente :** ici, le meilleur modèle est le plus simple, parce que les données synthétiques sont linéaires par construction. C'est une illustration du principe « la simplicité est une référence », pas un classement général. Sur des données réelles (partie B), il faudra refaire cette comparaison avec validation croisée.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Régression logistique (*Logistic Regression*). **Famille :** classification supervisée. **Objectif :** prédire la probabilité qu'une observation appartienne à la classe 1, puis en déduire une classe.

**Fonctionnement :** en entrée, $X$ (variables) et $y$ (classe 0 ou 1). L'apprentissage calcule un score linéaire $z$ pour chaque observation, le transforme en probabilité avec une fonction en forme de S (la sigmoïde), et ajuste les coefficients pour que ces probabilités collent aux classes observées. En sortie, on obtient une probabilité, puis une classe si on la compare à un seuil (0,5 par défaut).

**Formule :** $p(x) = \sigma(z) = \frac{1}{1+e^{-z}}$ avec $z = \beta_0 + \beta^\top x$. Ici $p(x)$ est la probabilité de la classe 1, $\sigma$ la sigmoïde, $z$ le score linéaire, $\beta_0$ la constante, $\beta$ le vecteur des coefficients et $x$ le vecteur des variables. Les coefficients minimisent la perte logistique $-\frac{1}{n}\sum_i [y_i\ln p_i + (1-y_i)\ln(1-p_i)]$, où $n$ est le nombre d'observations.

**Paramètres / hyperparamètres :** paramètres appris : $\beta_0$ et $\beta$ ; hyperparamètres : `C` (inverse de la force de régularisation, 1,0 par défaut), `max_iter` (1000 ici) et le seuil de décision.

**Avantages :** donne des probabilités et des coefficients interprétables. **Limites :** la frontière entre classes est linéaire, donc le modèle échoue quand les classes sont imbriquées ; la standardisation est utile pour la régularisation.

### Schéma explicatif

```python
# Schéma 8 : la sigmoïde transforme le score z en probabilité
z = np.linspace(-6, 6, 200)
p = 1 / (1 + np.exp(-z))
plt.figure(figsize=(6, 4))
plt.plot(z, p, color='tab:red', label='p = 1 / (1 + e^(-z))')
plt.axhline(.5, color='gray', ls='--', label='seuil de décision 0,5')
plt.axvline(0, color='gray', lw=.8)
plt.scatter([0, 2], [.5, 1 / (1 + np.exp(-2))], color='black', zorder=3)
plt.annotate('z = 0  →  p = 0,5', (0, .5), textcoords='offset points', xytext=(10, -18))
plt.annotate('z = 2  →  p ≈ 0,881', (2, 1 / (1 + np.exp(-2))), textcoords='offset points', xytext=(10, -18))
plt.xlabel('score linéaire z'); plt.ylabel('probabilité de la classe 1')
plt.title('Régression logistique : du score à la probabilité')
plt.legend(loc='upper left'); plt.show()
```

**Comment lire le schéma :** l'axe horizontal est le score linéaire $z$ calculé par le modèle, et l'axe vertical la probabilité d'être de la classe 1. Un score de 0 donne une probabilité de 0,5 ; plus $z$ est grand, plus la probabilité approche 1, et plus $z$ est négatif, plus elle approche 0. La ligne pointillée est le seuil : au-dessus, on prédit la classe 1. C'est un schéma du principe (une fonction), il ne vient pas des données.

### Contextes d'utilisation

- **Cas 1 (finance) :** estimer la probabilité de défaut de paiement d'un client à partir de son âge, de ses revenus et de son historique. **Cas 2 (marketing) :** prévoir si un prospect va répondre à une campagne.
- **Données et prérequis :** variables numériques ou encodées, cible binaire, variables standardisées si on régularise ; on choisit le seuil selon le coût des erreurs.
- **Cas peu approprié :** quand les classes sont séparées par une frontière très courbe ou imbriquées (sans transformation des variables, la frontière linéaire échoue).
- **Métriques :** F1, précision, rappel, ROC-AUC ; PR-AUC si la classe à détecter est rare.

### Code fourni (conservé) et sortie

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xc_ent` (300 lignes × 6 variables) et `yc_ent` pour l'apprentissage, `Xc_test` (100 lignes) pour l'évaluation ; le découpage est stratifié, donc le test contient 50 observations de chaque classe. Les données viennent de `make_classification` : elles sont **synthétiques**. **Hyperparamètres :** `max_iter=1000`, `C` par défaut (1,0), standardisation dans le `Pipeline`. **Sortie :** un F1 de 0,811 sur le test.

**Ce que la sortie permet de dire :** le F1 combine précision et rappel, et 0,811 montre un modèle correct mais plus faible que plusieurs modèles non linéaires de la suite (0,89 à 0,96). Une explication possible est que `make_classification` place par défaut deux groupes de points par classe, ce qui rend la frontière non linéaire ; c'est une hypothèse, que je n'ai pas testée.

**Ce qu'elle ne permet pas de dire :** le F1 est donné pour le seuil 0,5 seulement, sans ROC-AUC ni probabilités affichées. Le test ne contient que 100 observations : une erreur de plus ou de moins déplace le F1 d'environ un point. Il n'y a qu'un seul découpage et aucun réglage de `C`.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** K plus proches voisins (*k-Nearest Neighbors*, KNN). **Famille :** classification supervisée. **Objectif :** classer une observation d'après la classe majoritaire de ses voisins les plus proches.

**Fonctionnement :** en entrée, les observations d'entraînement avec leurs classes. Il n'y a pas vraiment d'apprentissage : le modèle mémorise les données. Pour classer un nouvel exemple, on mesure sa distance à tous les points d'entraînement, on retient les $k$ plus proches et on prend leur classe majoritaire.

**Formule :** $\hat{y}(x) = \text{mode}\{y_i : i \in N_k(x)\}$, avec $d(x,z) = \sqrt{\sum_j (x_j - z_j)^2}$. Ici $N_k(x)$ est l'ensemble des $k$ voisins de $x$, $y_i$ la classe du voisin $i$, $d$ la distance euclidienne, et $x_j$, $z_j$ les valeurs de la variable $j$ pour les deux points comparés.

**Paramètres / hyperparamètres :** il n'y a pas de paramètres appris (apprentissage « paresseux »). L'hyperparamètre principal est $k$ (`n_neighbors=5` ici) ; la distance et le poids des voisins en sont d'autres.

**Avantages :** très simple et capable de frontières non linéaires. **Limites :** sensible à l'échelle des variables (d'où la standardisation) et à la dimension ; la prédiction devient lente avec beaucoup de données.

### Schéma explicatif

```python
# Schéma 9 : frontière de décision du KNN avec k = 1 puis k = 15 (données jouet en deux demi-lunes)
from sklearn.datasets import make_moons
from sklearn.neighbors import KNeighborsClassifier
Xm, ym = make_moons(n_samples=200, noise=0.25, random_state=GRAINE)

def tracer_frontiere(modele, ax, titre):
    xx, yy = np.meshgrid(np.linspace(-2, 3, 250), np.linspace(-1.5, 2, 250))
    zz = modele.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, zz, alpha=.25, cmap='coolwarm')
    ax.scatter(Xm[:, 0], Xm[:, 1], c=ym, cmap='coolwarm', edgecolor='k', s=18)
    ax.set_title(titre); ax.set_xlabel('x1'); ax.set_ylabel('x2')

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for ax, k in zip(axes, [1, 15]):
    mod = KNeighborsClassifier(n_neighbors=k).fit(Xm, ym)
    tracer_frontiere(mod, ax, 'KNN, k = %d (précision sur ces points : %.2f)' % (k, mod.score(Xm, ym)))
plt.show()
```

**Comment lire le schéma :** les points sont les observations (deux couleurs = deux classes) et les zones colorées indiquent la classe que le modèle prédirait à cet endroit. Avec $k=1$ (à gauche), chaque point est son propre plus proche voisin : la précision sur ces points d'apprentissage vaut 1,00 (elle est écrite dans le titre) et la frontière suit les moindres détails, y compris les points isolés au milieu de l'autre classe, signe de surapprentissage. Avec $k=15$ (à droite), la précision sur ces mêmes points est plus faible, mais la frontière est plus régulière et regroupe les points voisins. Le choix de $k$ règle donc le compromis entre complexité et stabilité. Les données de ce schéma sont des demi-lunes synthétiques, utilisées pour l'illustration seulement.

### Contextes d'utilisation

- **Cas 1 (marketing) :** rattacher un nouveau client à un profil (fidèle, occasionnel, à risque) en regardant les clients qui lui ressemblent le plus. **Cas 2 :** recommander un produit d'après les produits achetés par des clients proches.
- **Données et prérequis :** variables numériques standardisées, un nombre de variables modéré, et un jeu d'entraînement qui reste en mémoire.
- **Cas peu approprié :** un très grand jeu de données ou beaucoup de variables (calcul lent, distances peu informatives), ou des variables d'échelles très différentes non standardisées.
- **Métriques :** F1, précision, rappel, ROC-AUC (à partir des proportions de voisins).

### Code fourni (conservé) et sortie

```python
from sklearn.neighbors import KNeighborsClassifier
modele=make_pipeline(StandardScaler(),KNeighborsClassifier(n_neighbors=5)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** les mêmes `Xc_ent` / `yc_ent` et `Xc_test`. Le `Pipeline` standardise d'abord, avec la moyenne et l'écart-type de l'entraînement seulement, puis applique KNN. **Hyperparamètre :** `n_neighbors=5`, non réglé. **Sortie :** un F1 de 0,889 sur le test.

**Ce que la sortie permet de dire :** sans phase d'apprentissage, KNN obtient un score nettement supérieur à celui de la régression logistique (0,811), ce qui va dans le sens d'une frontière non linéaire.

**Ce qu'elle ne permet pas de dire :** `make_classification` ajoute 2 variables sans information sur les 6, et elles perturbent les distances ; je n'ai pas mesuré l'effet. La valeur de $k$ n'est pas réglée, il n'y a qu'un seul découpage et un test de 100 observations, donc les écarts de quelques points entre modèles ne sont pas significatifs.

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Naive Bayes gaussien (*Gaussian Naive Bayes*). **Famille :** classification supervisée. **Objectif :** classer une observation avec la règle de Bayes, en supposant que les variables sont indépendantes entre elles à l'intérieur de chaque classe.

**Fonctionnement :** en entrée, $X$ et $y$. L'apprentissage estime, pour chaque classe, sa fréquence et la moyenne et la variance de chaque variable. Pour une nouvelle observation, on calcule pour chaque classe un score qui combine la fréquence de la classe et la vraisemblance de chaque variable ; la classe au score le plus élevé est choisie.

**Formule :** $P(C_k \mid x) \propto P(C_k)\prod_{j=1}^{p} P(x_j \mid C_k)$, avec $P(x_j \mid C_k) = \mathcal{N}(x_j ; \mu_{jk}, \sigma_{jk}^2)$. Ici $C_k$ est la classe $k$, $P(C_k)$ sa fréquence a priori, $x_j$ la variable $j$, $p$ le nombre de variables, $\mu_{jk}$ et $\sigma_{jk}^2$ la moyenne et la variance de la variable $j$ dans la classe $k$.

**Paramètres / hyperparamètres :** paramètres appris : les fréquences $P(C_k)$, les $\mu_{jk}$ et les $\sigma_{jk}^2$ ; hyperparamètre : `var_smoothing` (valeur par défaut ici), qui évite des variances nulles.

**Avantages :** très rapide et efficace même avec peu de données. **Limites :** l'indépendance entre variables est rarement vraie, et l'hypothèse gaussienne peut être fausse.

### Schéma explicatif

```python
# Schéma 10 : pour une variable, une gaussienne par classe (données synthétiques de l'entraînement)
from scipy.stats import norm
colonne = Xc_ent[:, 0]
grille = np.linspace(colonne.min(), colonne.max(), 200)
plt.figure(figsize=(6.5, 4))
for classe, couleur in [(0, 'tab:blue'), (1, 'tab:red')]:
    valeurs = colonne[yc_ent == classe]
    plt.hist(valeurs, bins=15, density=True, alpha=.3, color=couleur)
    plt.plot(grille, norm.pdf(grille, valeurs.mean(), valeurs.std()), color=couleur,
             label='classe %d : gaussienne (moyenne %.2f, écart-type %.2f)' % (classe, valeurs.mean(), valeurs.std()))
plt.xlabel('variable 1'); plt.ylabel('densité')
plt.title('Naive Bayes : une loi normale estimée par classe')
plt.legend(fontsize=8); plt.show()
```

**Comment lire le schéma :** pour la première variable, les histogrammes pâles montrent les valeurs observées dans chaque classe, et les courbes sont les lois normales que Naive Bayes estime à partir d'elles. Pour une nouvelle valeur, le modèle regarde la hauteur de chaque courbe (la vraisemblance) et la multiplie par la fréquence de la classe. Il fait la même chose pour **chaque variable séparément** puis multiplie les résultats : c'est l'hypothèse « naïve ». Plus les deux courbes se chevauchent, moins la variable aide à séparer les classes.

### Contextes d'utilisation

- **Cas 1 (commerce) :** filtrer les messages commerciaux indésirables ou classer des commentaires clients (positif/négatif) à partir de la présence de mots. **Cas 2 :** un diagnostic rapide à partir de quelques mesures, comme point de départ.
- **Données et prérequis :** variables numériques (version gaussienne) ou comptages de mots (autres versions), et un jeu d'entraînement même petit.
- **Cas peu approprié :** des variables fortement corrélées entre elles, ou dont les valeurs ne suivent pas du tout une loi normale par classe.
- **Métriques :** F1, précision, rappel, ROC-AUC ; ses probabilités sont souvent mal calibrées.

### Code fourni (conservé) et sortie

```python
from sklearn.naive_bayes import GaussianNB
modele=GaussianNB().fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xc_ent` / `yc_ent` pour apprendre les paramètres, `Xc_test` pour l'évaluation. Pas de standardisation, inutile ici. **Hyperparamètres :** aucun modifié. **Sortie :** un F1 de 0,755 sur le test.

**Ce que la sortie permet de dire :** c'est le score le plus bas des neuf classificateurs de ce notebook. Une explication plausible est que `make_classification` fabrique des variables liées entre elles et des classes en plusieurs groupes, ce qui contredit l'indépendance et la loi normale supposées ; c'est une hypothèse de ma part, non vérifiée.

**Ce qu'elle ne permet pas de dire :** on ne peut pas en conclure que Naive Bayes est un mauvais algorithme : sur des données où ses hypothèses sont à peu près vraies (par exemple des mots dans des textes), il est souvent très compétitif. Le résultat dépend de ces données synthétiques, d'un seul découpage et de 100 observations de test.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Arbre de décision pour la classification (*Decision Tree Classifier*). **Famille :** classification supervisée. **Objectif :** classer une observation grâce à une suite de questions simples sur ses variables.

**Fonctionnement :** en entrée, $X$ et $y$. À chaque nœud, l'algorithme essaie des coupures du type « variable $j$ inférieure à un seuil » et garde celle qui rend les classes les plus pures dans les deux groupes obtenus. Il recommence dans chaque groupe jusqu'à un critère d'arrêt. En sortie, une observation descend dans l'arbre jusqu'à une feuille, qui donne la classe majoritaire de cette feuille.

**Formule :** $G(t) = 1 - \sum_{k=1}^{K} p_{k,t}^2$ (impureté de Gini). Ici $t$ est un nœud, $K$ le nombre de classes et $p_{k,t}$ la proportion d'observations de la classe $k$ dans le nœud $t$. Un nœud à 50 % / 50 % a $G = 0{,}5$ ; un nœud pur a $G = 0$.

**Paramètres / hyperparamètres :** paramètres appris : les variables et seuils de coupure et la classe de chaque feuille ; hyperparamètres : `max_depth` (4 ici), `min_samples_leaf`, etc.

**Avantages :** règles faciles à lire et à expliquer, sans standardisation. **Limites :** instable (un petit changement des données peut changer l'arbre) et il surapprend quand il est profond.

### Schéma explicatif

```python
# Schéma 11 : un arbre de décision de profondeur 2 appris sur l'entraînement
from sklearn.tree import DecisionTreeClassifier, plot_tree
arbre2 = DecisionTreeClassifier(max_depth=2, random_state=GRAINE).fit(Xc_ent, yc_ent)
plt.figure(figsize=(9, 4.5))
plot_tree(arbre2, feature_names=['x%d' % (j + 1) for j in range(6)],
          class_names=['classe 0', 'classe 1'], filled=True, fontsize=9)
plt.title('Arbre de décision (profondeur 2) : questions successives et impureté de Gini')
plt.show()
```

**Comment lire le schéma :** on part du nœud du haut (la racine) : il pose une question sur une variable (dans mon exécution, « x6 ≤ 0,568 »). Si la réponse est oui, on suit la branche de gauche, sinon celle de droite. Dans chaque case, `gini` est l'impureté du nœud, `samples` le nombre d'observations qui y arrivent, `value` le nombre d'observations de chaque classe et `class` la classe majoritaire. La couleur est plus foncée quand le nœud est plus pur. Les feuilles tout en bas donnent la prédiction. Les données sont les données synthétiques de l'entraînement.

### Contextes d'utilisation

- **Cas 1 (commerce international) :** décider si un marché export est « à qualifier » d'après des règles simples (taille du marché, droits de douane, risque pays). **Cas 2 (gestion) :** accorder ou refuser un crédit avec des règles que le conseiller peut expliquer au client.
- **Données et prérequis :** variables numériques ou catégorielles encodées, pas de standardisation, et un contrôle de la profondeur pour éviter le surapprentissage.
- **Cas peu approprié :** quand la frontière entre classes est lisse et oblique (l'arbre la découpe en escaliers), ou quand une petite variation des données ne doit pas changer le modèle.
- **Métriques :** F1, précision, rappel, ROC-AUC ; lisibilité des règles.

### Code fourni (conservé) et sortie

```python
from sklearn.tree import DecisionTreeClassifier
modele=DecisionTreeClassifier(max_depth=4,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xc_ent` / `yc_ent` et `Xc_test`. **Hyperparamètres :** `max_depth=4` et `random_state=GRAINE`, non réglés. **Sortie :** un F1 de 0,891 sur le test.

**Ce que la sortie permet de dire :** un seul arbre limité à 4 niveaux fait beaucoup mieux que la régression logistique (0,811), ce qui montre que des coupures successives suffisent à séparer ces classes. Il est proche de KNN (0,889) et du SVM (0,898).

**Ce qu'elle ne permet pas de dire :** la profondeur 4 n'a pas été choisie sur une validation. Avec 100 observations de test, un écart de 1 à 3 points de F1 correspond à quelques observations seulement : on ne peut pas classer ces trois modèles entre eux. Les données sont synthétiques.

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Forêt aléatoire pour la classification (*Random Forest Classifier*). **Famille :** classification supervisée (méthode d'ensemble). **Objectif :** classer en faisant voter de nombreux arbres différents, pour être plus stable qu'un arbre seul.

**Fonctionnement :** en entrée, $X$ et $y$. L'apprentissage construit chaque arbre sur un échantillon tiré avec remise (bootstrap), et à chaque coupure l'arbre ne regarde qu'un sous-ensemble aléatoire de variables. En sortie, chaque arbre vote et la classe qui a le plus de votes est retenue.

**Formule :** $\hat{y}(x) = \text{mode}\{T_1(x), \dots, T_B(x)\}$. Ici $B$ est le nombre d'arbres, $T_b(x)$ la classe prédite par l'arbre $b$ pour l'entrée $x$, et « mode » la classe la plus fréquente parmi les votes.

**Paramètres / hyperparamètres :** paramètres appris : les coupures et feuilles de chaque arbre ; hyperparamètres : `n_estimators` (80 ici), `max_features` (par défaut, la racine carrée du nombre de variables pour la classification), `max_depth`.

**Avantages :** bonne précision sur des données tabulaires et peu de réglages. **Limites :** il est plus difficile d'expliquer une prédiction individuelle qu'avec un arbre, et le modèle est volumineux.

### Schéma explicatif

```python
# Schéma 12 : un arbre profond seul contre une forêt (données jouet en deux demi-lunes)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
arbre_libre = DecisionTreeClassifier(random_state=GRAINE).fit(Xm, ym)
foret = RandomForestClassifier(n_estimators=80, random_state=GRAINE).fit(Xm, ym)
tracer_frontiere(arbre_libre, axes[0], 'Un arbre sans limite (profondeur %d, %d feuilles)' % (arbre_libre.get_depth(), arbre_libre.get_n_leaves()))
tracer_frontiere(foret, axes[1], 'Forêt de 80 arbres')
plt.show()
print("Précision sur les points d'apprentissage : arbre", arbre_libre.score(Xm, ym), '| forêt', foret.score(Xm, ym))
```

**Comment lire le schéma :** à gauche, un seul arbre sans limite de profondeur : il se développe jusqu'à retrouver tous les points d'apprentissage (la précision sur ces points est affichée sous le schéma), y compris des points isolés au milieu de l'autre classe, ce qui annonce un risque de surapprentissage. À droite, la forêt combine 80 arbres entraînés sur des échantillons différents. Les deux frontières se ressemblent dans les grandes lignes, mais celle de la forêt est plus nuancée, parce que les 80 arbres ne coupent pas aux mêmes endroits et votent ensemble. Les données sont les demi-lunes synthétiques du schéma 9.

### Contextes d'utilisation

- **Cas 1 (finance / commerce) :** détecter un risque commercial ou un défaut de paiement à partir de nombreuses variables sur le client. **Cas 2 :** prévoir le désabonnement d'un client à partir de son historique d'usage.
- **Données et prérequis :** données tabulaires numériques ou encodées, pas de standardisation, et un nombre d'arbres suffisant.
- **Cas peu approprié :** une décision qui doit être expliquée règle par règle à chaque client, ou un contexte avec très peu de mémoire et besoin d'un modèle léger.
- **Métriques :** F1, précision, rappel, ROC-AUC ; erreur « out-of-bag » pour estimer la généralisation.

### Code fourni (conservé) et sortie

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Machine à vecteurs de support (*Support Vector Machine*, SVM). **Famille :** classification supervisée. **Objectif :** trouver la frontière qui sépare les classes avec la plus grande marge possible.

**Fonctionnement :** en entrée, $X$ standardisée et $y$. L'apprentissage cherche l'hyperplan qui sépare les classes avec la marge la plus large, en tolérant quelques points mal placés. Avec le noyau RBF, la séparation se fait dans un espace transformé, ce qui donne une frontière courbe. En sortie, la classe dépend du signe de la fonction de décision.

**Formule :** $\min_{w,b,\xi}\ \frac{1}{2}\lVert w\rVert^2 + C\sum_i \xi_i$ avec $y_i(w^\top x_i + b) \ge 1 - \xi_i$. Ici $w$ est le vecteur normal à l'hyperplan, $b$ le biais, $\xi_i \ge 0$ l'écart de l'observation $i$ à la marge, $C$ la pénalité de ces écarts, $y_i \in \{-1, +1\}$ la classe et $x_i$ l'observation. La marge vaut $2/\lVert w\rVert$.

**Paramètres / hyperparamètres :** paramètres appris : $w$ et $b$ (ou les vecteurs de support et leurs poids avec un noyau) ; hyperparamètres : `C` (1,0 ici), `kernel` (`rbf`) et `gamma` (valeur par défaut `scale`).

**Avantages :** efficace en dimension moyenne et capable de frontières non linéaires grâce aux noyaux. **Limites :** coûteux sur de grands jeux de données, sensible au choix de $C$, $\gamma$ et à l'échelle des variables, et sans probabilités natives.

### Schéma explicatif

```python
# Schéma 13 : SVM à noyau RBF, marge et vecteurs de support (données jouet en demi-lunes)
from sklearn.svm import SVC
svm = SVC(kernel='rbf', C=1.0, gamma='scale').fit(Xm, ym)
xx, yy = np.meshgrid(np.linspace(-2, 3, 250), np.linspace(-1.5, 2, 250))
dec = svm.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
plt.figure(figsize=(6.5, 4.5))
plt.contourf(xx, yy, dec > 0, alpha=.25, cmap='coolwarm')
plt.contour(xx, yy, dec, levels=[-1, 0, 1], colors='k', linestyles=['--', '-', '--'])
plt.scatter(Xm[:, 0], Xm[:, 1], c=ym, cmap='coolwarm', edgecolor='k', s=18)
plt.scatter(svm.support_vectors_[:, 0], svm.support_vectors_[:, 1], s=110,
            facecolors='none', edgecolors='k', linewidths=1.2, label='vecteurs de support')
plt.xlabel('x1'); plt.ylabel('x2'); plt.title('SVM RBF : frontière, marge et vecteurs de support')
plt.legend(loc='upper right'); plt.show()
```

**Comment lire le schéma :** le trait plein est la frontière de décision ; les deux traits pointillés délimitent la marge de part et d'autre. Les points entourés sont les **vecteurs de support** : ce sont les seuls points qui déterminent la frontière (ceux qui sont sur la marge ou du mauvais côté). Les autres points pourraient être retirés sans changer le modèle. Le noyau RBF permet à cette frontière d'être courbe. Les données sont des demi-lunes synthétiques, utilisées pour l'illustration.

### Contextes d'utilisation

- **Cas 1 (finance) :** classer des opérations rares à risque à partir de quelques indicateurs bien choisis, quand on a quelques milliers d'observations. **Cas 2 :** reconnaissance de formes ou classification de textes courts représentés par des vecteurs.
- **Données et prérequis :** variables numériques standardisées, jeu de taille petite à moyenne, et réglage de $C$ et de $\gamma$ par validation croisée.
- **Cas peu approprié :** de très grands jeux de données (le temps d'apprentissage augmente fortement), ou un besoin de probabilités directement fiables.
- **Métriques :** F1, précision, rappel, ROC-AUC (à partir de la fonction de décision).

### Code fourni (conservé) et sortie

```python
from sklearn.svm import SVC
modele=make_pipeline(StandardScaler(),SVC(kernel='rbf',C=1.0,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xc_ent` / `yc_ent` et `Xc_test`, avec standardisation dans le `Pipeline` (indispensable pour le noyau RBF). **Hyperparamètres :** noyau `rbf`, `C=1.0`, `gamma` par défaut, rien n'est réglé. **Sortie :** un F1 de 0,898 sur le test.

**Ce que la sortie permet de dire :** le SVM RBF est nettement meilleur que la régression logistique (0,811), ce qui montre qu'une frontière courbe aide sur ces données. Il est au même niveau que l'arbre (0,891) et KNN (0,889).

**Ce qu'elle ne permet pas de dire :** `SVC` est ici créé sans `probability=True` (le cours l'indique), donc il n'y a pas de probabilités ; pour un ROC-AUC je devrai utiliser la fonction de décision. Les hyperparamètres ne sont pas réglés et les trois scores voisins ne se distinguent pas avec 100 observations de test.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** Gradient boosting pour la classification (*Gradient Boosting Classifier*) ; l'exemple utilise `HistGradientBoostingClassifier`. **Famille :** classification supervisée (méthode d'ensemble séquentielle). **Objectif :** classer en ajoutant des petits arbres qui corrigent progressivement les erreurs du modèle.

**Fonctionnement :** en entrée, $X$ et $y$. On commence par un score simple, puis à chaque étape on calcule dans quelle direction le score doit bouger pour réduire la perte logistique, on entraîne un petit arbre sur cette correction et on l'ajoute au modèle, réduit par un pas. En sortie, le score final est transformé en probabilité, puis en classe.

**Formule :** $F_m(x) = F_{m-1}(x) + \eta\, h_m(x)$, avec $p(x) = \sigma(F_M(x))$. Ici $F_m$ est le score après $m$ étapes, $h_m$ le nouvel arbre, $\eta$ le taux d'apprentissage, $M$ le nombre total d'étapes et $\sigma$ la sigmoïde (cas à deux classes).

**Paramètres / hyperparamètres :** paramètres appris : les arbres ; hyperparamètres : `max_iter` (nombre d'arbres, 80 ici), `learning_rate` ($\eta$, 0,1 par défaut) et la complexité de chaque arbre. La version « Hist » regroupe les valeurs des variables en classes pour être plus rapide.

**Avantages :** très compétitif sur les données tabulaires. **Limites :** risque de surapprentissage si on ajoute trop d'arbres, et réglages plus sensibles que ceux d'une forêt.

### Schéma explicatif

```python
# Schéma 14 : la frontière du boosting s'affine quand on ajoute des arbres (données jouet)
from sklearn.ensemble import HistGradientBoostingClassifier
fig, axes = plt.subplots(1, 3, figsize=(13, 4))
for ax, nb_arbres in zip(axes, [1, 10, 100]):
    mod = HistGradientBoostingClassifier(max_iter=nb_arbres, learning_rate=.3, min_samples_leaf=5,
                                         max_leaf_nodes=8, random_state=GRAINE).fit(Xm, ym)
    tracer_frontiere(mod, ax, '%d arbre(s) | précision apprentissage %.2f' % (nb_arbres, mod.score(Xm, ym)))
plt.show()
```

**Comment lire le schéma :** de gauche à droite, le même modèle avec 1, 10 puis 100 arbres. Avec un seul arbre, la frontière est grossière (quelques rectangles) ; chaque arbre ajouté corrige une partie des erreurs restantes, donc la frontière se précise et la précision sur les points d'apprentissage, écrite dans les titres, passe de 0,96 à 0,98 puis à 1,00. Atteindre 100 % sur les points d'apprentissage alors que les deux classes se chevauchent signifie que le modèle commence à mémoriser le bruit : c'est un risque de surapprentissage. Le nombre d'arbres est donc un hyperparamètre à régler sur des données de validation. Les données sont synthétiques (schéma 9).

### Contextes d'utilisation

- **Cas 1 (marketing) :** prévoir si un client va souscrire à une offre à partir de son profil et de ses interactions passées. **Cas 2 (finance) :** score de risque de crédit sur des données tabulaires riches.
- **Données et prérequis :** données tabulaires de taille moyenne à grande, validation soignée pour régler le nombre d'arbres et $\eta$.
- **Cas peu approprié :** un très petit jeu de données (surapprentissage), ou une décision réglementaire qui exige une explication simple.
- **Métriques :** F1, précision, rappel, ROC-AUC, PR-AUC si la classe positive est rare.

### Code fourni (conservé) et sortie

```python
from sklearn.ensemble import HistGradientBoostingClassifier
modele=HistGradientBoostingClassifier(max_iter=80,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xc_ent` / `yc_ent` et `Xc_test`. **Hyperparamètres :** `max_iter=80` et `random_state=GRAINE`, les autres valeurs sont celles par défaut (dont `learning_rate=0.1`). **Sortie :** un F1 de 0,960 sur le test dans mon exécution Colab.

**Ce que la sortie permet de dire :** c'est le meilleur F1 des neuf classificateurs, devant XGBoost (0,931) et la forêt (0,918). Cela va dans le sens de l'idée que le boosting est très compétitif sur des tableaux de données.

**Ce qu'elle ne permet pas de dire :** avec 100 observations de test, un écart de 3 à 4 points correspond à quelques observations seulement, donc on ne peut pas affirmer que ce modèle est supérieur aux autres. Le nombre d'itérations n'est pas choisi sur une validation. **Remarque :** le résultat d'un modèle de boosting peut varier légèrement d'une version de scikit-learn à l'autre, c'est pourquoi la cellule des versions est utile pour reproduire cette valeur.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** AdaBoost (*Adaptive Boosting*). **Famille :** classification supervisée (méthode d'ensemble séquentielle). **Objectif :** construire un classificateur fort en combinant beaucoup de classificateurs faibles, en insistant à chaque étape sur les observations mal classées.

**Fonctionnement :** en entrée, $X$ et $y$. Toutes les observations reçoivent d'abord le même poids. On entraîne un classificateur très simple (par défaut un arbre à une seule coupure), on augmente le poids des observations qu'il a mal classées, puis on en entraîne un autre qui va donc se concentrer sur elles. En sortie, la décision est un vote pondéré de tous les classificateurs.

**Formule :** $F(x) = \sum_{m=1}^{M} \alpha_m h_m(x)$. Ici $M$ est le nombre de classificateurs faibles, $h_m(x)$ la prédiction du classificateur $m$, et $\alpha_m$ son poids dans le vote : il est d'autant plus grand que l'erreur pondérée $\varepsilon_m$ du classificateur est faible (dans la version classique, $\alpha_m = \frac{1}{2}\ln\frac{1-\varepsilon_m}{\varepsilon_m}$ ; scikit-learn utilise une variante, SAMME, de même logique).

**Paramètres / hyperparamètres :** paramètres appris : les classificateurs et leurs poids $\alpha_m$ ; hyperparamètres : `n_estimators` (70 ici) et `learning_rate`.

**Avantages :** simple à mettre en œuvre avec des classificateurs faibles et réduit le biais d'un modèle trop simple. **Limites :** très sensible au bruit et aux étiquettes erronées, car il leur donne de plus en plus de poids.

### Schéma explicatif

```python
# Schéma 15 : erreur pondérée et poids de chacun des classificateurs faibles d'AdaBoost
from sklearn.ensemble import AdaBoostClassifier
ada = AdaBoostClassifier(n_estimators=70, random_state=GRAINE).fit(Xc_ent, yc_ent)
etapes = np.arange(1, len(ada.estimator_errors_) + 1)
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
axes[0].plot(etapes, ada.estimator_errors_, color='tab:blue')
axes[0].set_title("Erreur pondérée de chaque classificateur faible")
axes[0].set_xlabel("étape m"); axes[0].set_ylabel("erreur pondérée")
axes[1].bar(etapes, ada.estimator_weights_, color='tab:orange')
axes[1].set_title("Poids de chaque classificateur dans le vote final")
axes[1].set_xlabel("étape m"); axes[1].set_ylabel("poids")
plt.show()
```

**Comment lire le schéma :** à gauche, l'erreur pondérée de chaque classificateur faible, étape par étape. À droite, le poids que chacun reçoit dans le vote final. Un classificateur qui se trompe peu a un poids plus fort. Les classificateurs suivants sont entraînés sur des observations de plus en plus difficiles (les poids des observations sont réajustés), donc leur erreur sur ces observations tend à s'élever à mesure que les étapes avancent. Les données sont les données synthétiques de l'entraînement du notebook.

### Contextes d'utilisation

- **Cas 1 (marketing) :** classer des prospects (intéressé / pas intéressé) avec des règles très simples combinées. **Cas 2 :** détection de visages en vision par ordinateur, un usage historique d'AdaBoost.
- **Données et prérequis :** données tabulaires propres, avec des étiquettes fiables ; on règle le nombre d'estimateurs.
- **Cas peu approprié :** des données avec beaucoup d'étiquettes erronées ou de valeurs aberrantes, car l'algorithme accentue leur poids.
- **Métriques :** F1, précision, rappel, ROC-AUC.

### Code fourni (conservé) et sortie

```python
from sklearn.ensemble import AdaBoostClassifier
modele=AdaBoostClassifier(n_estimators=70,random_state=GRAINE).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xc_ent` / `yc_ent` et `Xc_test`. **Hyperparamètres :** `n_estimators=70` et `random_state=GRAINE` ; le classificateur faible par défaut est un arbre à une coupure et `learning_rate` vaut 1,0. **Sortie :** un F1 de 0,827 sur le test.

**Ce que la sortie permet de dire :** c'est un résultat inférieur à ceux des autres méthodes d'ensemble (forêt 0,918, XGBoost 0,931, boosting 0,960), et proche de celui de la régression logistique (0,811). Une explication possible est que des arbres à une seule coupure sont trop simples pour cette structure en plusieurs groupes ; c'est une hypothèse que je n'ai pas testée (par exemple en augmentant la profondeur des classificateurs faibles).

**Ce qu'elle ne permet pas de dire :** on ne peut pas conclure qu'AdaBoost est moins bon que les autres boostings en général : le résultat dépend de l'arbre de base par défaut, de 70 itérations non réglées, de données synthétiques et d'un test de 100 observations.

## 16. XGBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation

**Nom :** XGBoost (*Extreme Gradient Boosting*). **Famille :** classification supervisée (méthode d'ensemble séquentielle). **Objectif :** une implémentation très optimisée du gradient boosting d'arbres, avec une pénalité explicite sur la complexité des arbres.

**Fonctionnement :** en entrée, $X$ et $y$. Comme pour le gradient boosting, on ajoute des arbres un par un, chacun corrigeant les erreurs restantes. XGBoost utilise en plus la courbure de la perte pour choisir les coupures et pénalise les arbres trop complexes. En sortie, on obtient une probabilité de classe, puis une classe.

**Formule :** $L = \sum_{i=1}^{n} \ell(y_i, \hat{y}_i) + \sum_{m=1}^{M} \Omega(f_m)$. Ici $\ell$ est la perte pour l'observation $i$ (entre la vraie valeur $y_i$ et la prédiction $\hat{y}_i$), $f_m$ le $m$-ième arbre, $M$ le nombre d'arbres et $\Omega$ la pénalité de complexité des arbres.

**Paramètres / hyperparamètres :** paramètres appris : les arbres et leurs feuilles ; hyperparamètres : `n_estimators` (80 ici), `max_depth` (3), `learning_rate` (valeur par défaut) et la régularisation. `eval_metric='logloss'` ne sert qu'à suivre la perte, il ne change pas les prédictions.

**Avantages :** très performant et rapide sur des tableaux de données, avec régularisation intégrée. **Limites :** une dépendance supplémentaire à installer, et beaucoup d'hyperparamètres à régler.

### Schéma explicatif

```python
# Schéma 16 : perte d'apprentissage et de validation de XGBoost au fil des arbres
from xgboost import XGBClassifier
X_a, X_v, y_a, y_v = train_test_split(Xc_ent, yc_ent, test_size=.25, stratify=yc_ent, random_state=GRAINE)
xg = XGBClassifier(n_estimators=80, max_depth=3, random_state=GRAINE, eval_metric='logloss')
xg.fit(X_a, y_a, eval_set=[(X_a, y_a), (X_v, y_v)], verbose=False)   # le test final n'est pas utilisé
resultats = xg.evals_result()
plt.figure(figsize=(6.5, 4))
plt.plot(resultats['validation_0']['logloss'], color='tab:blue', label='apprentissage')
plt.plot(resultats['validation_1']['logloss'], color='tab:red', label='validation')
plt.xlabel("nombre d'arbres"); plt.ylabel('perte logistique (logloss)')
plt.title("XGBoost : la perte selon le nombre d'arbres")
plt.legend(); plt.show()
```

**Comment lire le schéma :** chaque courbe montre la perte logistique après chaque arbre ajouté (plus elle est basse, mieux c'est). La courbe bleue est calculée sur les données d'apprentissage et la rouge sur un petit jeu de **validation** mis de côté dans l'entraînement ; le test final n'est pas utilisé. La perte d'apprentissage continue de baisser jusqu'au dernier arbre, alors que la perte de validation baisse au début puis **remonte** après une dizaine d'arbres environ : quand l'écart entre les deux grandit, le modèle commence à surapprendre et on devrait arrêter d'ajouter des arbres (arrêt anticipé). Avec un jeu de validation de seulement 75 observations, ce point d'arrêt est approximatif. Les données sont les données synthétiques du notebook.

### Contextes d'utilisation

- **Cas 1 (marketing) :** prévoir le désengagement d'un client (churn) à partir d'un grand nombre de variables d'usage. **Cas 2 (finance) :** détecter des opérations frauduleuses ou scorer un risque de crédit.
- **Données et prérequis :** données tabulaires, la bibliothèque `xgboost` installée, et une validation soignée (éventuellement avec arrêt anticipé) pour régler le nombre d'arbres et le pas.
- **Cas peu approprié :** un jeu de données très petit, ou une situation où un modèle simple et expliqué est exigé.
- **Métriques :** F1, précision, rappel, ROC-AUC et PR-AUC.

### Code fourni (conservé) et sortie

```python
from xgboost import XGBClassifier
modele=XGBClassifier(n_estimators=80,max_depth=3,random_state=GRAINE,eval_metric='logloss').fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

### Analyse de l'expérience Python

**Entrées :** `Xc_ent` / `yc_ent` et `Xc_test`. **Hyperparamètres :** `n_estimators=80`, `max_depth=3`, `random_state=GRAINE`, `eval_metric='logloss'` ; le pas et la régularisation gardent leurs valeurs par défaut. **Sortie :** un F1 de 0,931 sur le test.

**Ce que la sortie permet de dire :** XGBoost fait partie des modèles les plus performants (0,931), au niveau du boosting `Hist` (0,960) et de la forêt (0,918), et bien au-dessus de la régression logistique (0,811).

**Ce qu'elle ne permet pas de dire :** les écarts avec les autres méthodes d'ensemble sont de quelques observations sur 100 et ne sont pas significatifs. Aucun réglage n'a été fait (pas d'arrêt anticipé), et les données sont synthétiques.

## Comparaison des algorithmes de classification 8 à 16 (données synthétiques)

Le TAF demande F1, rappel, précision et ROC-AUC ou PR-AUC. Je recalcule ces métriques pour les 9 modèles avec le **même découpage train/test**, et je note le temps d'ajustement. Les classes sont équilibrées (50 / 50 dans le test), donc l'accuracy ne serait pas trompeuse ici, mais je garde F1 comme indicateur principal. Pour le SVM, qui n'a pas de probabilités dans ce code, j'utilise la fonction de décision pour calculer les AUC. Rappel : les données viennent de `make_classification`, ce tableau **n'est pas** un résultat sur des données réelles.

```python
import time
from sklearn.metrics import precision_score, recall_score, roc_auc_score, average_precision_score
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier, AdaBoostClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier

modeles_clf = {
    '8. Régression logistique': make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, random_state=GRAINE)),
    '9. KNN': make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5)),
    '10. Naive Bayes': GaussianNB(),
    '11. Arbre de décision': DecisionTreeClassifier(max_depth=4, random_state=GRAINE),
    '12. Forêt aléatoire': RandomForestClassifier(n_estimators=80, random_state=GRAINE),
    '13. SVM (RBF)': make_pipeline(StandardScaler(), SVC(kernel='rbf', C=1.0, random_state=GRAINE)),
    '14. Gradient boosting (Hist)': HistGradientBoostingClassifier(max_iter=80, random_state=GRAINE),
    '15. AdaBoost': AdaBoostClassifier(n_estimators=70, random_state=GRAINE),
    '16. XGBoost': XGBClassifier(n_estimators=80, max_depth=3, random_state=GRAINE, eval_metric='logloss'),
}
lignes = []
for nom, mod in modeles_clf.items():
    t0 = time.perf_counter()
    mod.fit(Xc_ent, yc_ent)
    duree = time.perf_counter() - t0
    pred = mod.predict(Xc_test)
    if hasattr(mod, 'predict_proba'):
        score = mod.predict_proba(Xc_test)[:, 1]
    else:
        score = mod.decision_function(Xc_test)
    lignes.append({'Algorithme': nom,
                   'Précision': precision_score(yc_test, pred),
                   'Rappel': recall_score(yc_test, pred),
                   'F1': f1_score(yc_test, pred),
                   'ROC-AUC': roc_auc_score(yc_test, score),
                   'PR-AUC': average_precision_score(yc_test, score),
                   'Temps ajustement (s)': duree})
tableau_clf = pd.DataFrame(lignes).round(3)
print(tableau_clf.to_string(index=False))
print('Répartition des classes dans le test :', np.bincount(yc_test))
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

## Annexes — implémentations manuelles des trois algorithmes imposés

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
